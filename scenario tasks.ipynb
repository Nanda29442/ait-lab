```python
from collections import deque
graph={
    "master":["E1","E2"],
    "E1":["E3","E4"],
    "E2":["E5"],
    "E3":[],
    "E4":[],
    "E5":[]
}
offline={"E4"}
visited=set()
queue=deque([("master",0)])
failed=[]
max_time=0
while queue:
  node,time=queue.popleft()
  if node in visited:
    continue
  visited.add(node)
  if node in offline:
    failed.append(node)
    continue
  print(f"{node} reached at {time}sec")
  max_time=max(max_time,time)
  for neighbor in graph[node]:
    if neighbor not in visited:
      queue.append((neighbor,time+1))
      print("Minimum Time:",max_time,"seconds")
      print("offline servers:",failed)
```

```python
graph={
    'A':['B','C'],
    'B':['A','B','C'],
    'C':['A','F'],
    'D':['B'],
    'E':['B','F'],
    'F':['C','F','G'],
    'G':['F']
}
visited=set()
path=[]
def dfs(node):
  visited.add(node)
  path.append(node)
  print("visiting:",node)
  for neighbor in graph[node]:
    if neighbor not in visited:
      dfs(neighbor)
  print("Backtracking from:",node)

start='A'
dfs(start)
print("Exploration path:", "--".join(path))
print("visited chambers:",visited)
```
