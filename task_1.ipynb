```python
from collections import deque
graph = {
    'A':['B','C'],
    'B':[],
    'C':[],

}
visited = set()
queue = deque()
start = 'A'
queue.append(start)
print("BFS traversal:")
while queue:
    node = queue.popleft()
    if node not in visited:
        print(node,end=" ")
        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                queue.append(neighbor)
```

```python
graph = {
    'A': ['B', 'C'],
    'B': [],
    'C': [],
}

visited = set()
stack = []
start = 'A'

stack.append(start)

print("DFS traversal:")

while stack:
    node = stack.pop()

    if node not in visited:
        print(node, end=" ")
        visited.add(node)
        for neighbor in reversed(graph[node]):
            if neighbor not in visited:
                stack.append(neighbor)
```
