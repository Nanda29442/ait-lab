```python
import heapq


def h(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])


def a_star(grid, start, goal):
    q = [(h(start, goal), 0, start, [start])]

    cost = {start: 0}

    while q:
        f, g, p, path = heapq.heappop(q)

        if p == goal:
            return path, g

        for dx, dy in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
            x, y = p[0] + dx, p[1] + dy


            if 0 <= x < len(grid) and 0 <= y < len(grid[0]):
                if grid[x][y] == 1:
                    continue

                ng = g + 1
                new_position = (x, y)

                if new_position not in cost or ng < cost[new_position]:
                    cost[new_position] = ng

                    f_cost = ng + h(new_position, goal)

                    heapq.heappush(
                        q,
                        (f_cost, ng, new_position, path + [new_position])
                    )

    return None, float("inf")
grid = [
    [0, 0, 0],
    [0, 1, 0],
    [0, 0, 0]
]

start = (0, 0)
goal = (2, 2)

print("Input grid:")
for row in grid:
    print(row)

print("Start:", start)
print("Goal:", goal)

path, cost = a_star(grid, start, goal)

print("Optimal path:", path)
print("Cost:", cost)
```
